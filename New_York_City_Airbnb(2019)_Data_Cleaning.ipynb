```python
import kagglehub
import os
import pandas as pd
```

```python
path = kagglehub.dataset_download("dgomonov/new-york-city-airbnb-open-data")

print("Path to dataset files:", path)
```

```python
files= os.listdir(path)
print('Files in Directory: ',files)
```

```python
csv_file= [f for f in files if f.endswith(".csv")][0]
data_file=os.path.join(path,csv_file)
```

```python
df= pd.read_csv(data_file)
pd.set_option('display.max_columns', None)

# Increase the display width limit to prevent wrapping
pd.set_option('display.width', 1000)

print(df)
```

```python
#handling missing values
df.loc[df['number_of_reviews'] == 0, 'reviews_per_month'] = 0.0
```

```python
#Handles Readable Name
df=df.rename(columns={
    'name': 'AirBnB',
    'neighbourhood_group':'Borough',
    'availability_365':'availability'
})
#Handles symbols and formats
text_columns=df.select_dtypes(include=['object']).columns
for col in text_columns:
    df[col]=(
        df[col].fillna('unkown')
        .astype(str)
        .str.lower()
        .str.replace('[^\w\s]','',regex=True)
        .str.strip()
        .str.title()
    )
    print('All Text column is standarized and symbol free...')
```

```python
#Handles NaN datas
df['last_review']=pd.to_datetime(df['last_review'],errors='coerce')
df['AirBnB']=df['AirBnB'].fillna('unknown')
df['host_name']=df['host_name'].fillna('unknown host')
```

```python
print(df)
```

```python
#checking for NaN's
print(df.isna().sum())
```
